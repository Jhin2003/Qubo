```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Load embedding model (you can change this)
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

# ---- Step 1: Define your domain description ----
domain_description = """
Philippine cultural history including pre-colonial traditions,
indigenous groups, Spanish and American colonial influences,
Filipino customs, arts, literature, and heritage.
"""

# ---- Step 2: Sample document chunks ----
# Sample chunks extracted from your Topic 1 Pre-16 Century Philippines readings
chunks = [
    """Topic 1: Pre-16 Century Philippines Reading Materials
    1. “A New Species of Homo in the Late Pleistocene of the Philippines”
    2. “Upper Pleistocene Homo sapiens From the Tabon Cave (Palawan, The Philippines): Description and Dating of New Discoveries”
    3. “The Austronesians: Historical and Comparative Perspective”
    4. “Soul Boats: A Filipino Journey of Self-Discovery”
    5. CHAO JU-KUA’S DESCRIPTION OF THE PHILIPPINES
    6. “Questions and Challenges in Philippine Prehistory”
    7. Kalantiaw: The Code that Never Was
    8. The First People of Sulu""",

    """The first excerpt is from the article of archaeologists Armand Salvador Mijares and his team discussing the initial finds in Callao Cave in Peñablanca, Cagayan. These findings helped establish the human remains as Homo luzonensis, contributing to the Philippines' place in the archaeological map of human evolution.""",

    """The second excerpt is from archaeologist Eusebio Dizon and his team on the human remains from the Tabon Cave in Palawan. Discovered in the 1960s, recent reassessment yielded new dates and eleven additional human remains.""",

    """Readings three, four and five are from the study of Peter Bellwood regarding the Austronesians. These excerpts discuss the migration and settlement of Austronesians in the Philippines, Southeast Asia, the Pacific Islands, and Madagascar.""",

    """Reading six is from Alfredo E. Evangelista on ancient Filipino life using 'soul boats,' wooden coffins shaped like boats found in archaeological sites in the Philippines.""",

    """Reading seven is CHAO JU-KUA’S document detailing contacts with the peoples of Ma-yi, recognized as either Manila or Mindoro.""",

    """Readings eight and nine are studies on methodological problems in Philippine Prehistory. Reading eight is F. Landa Jocano on research challenges, and reading nine is William Henry Scott on the Maragtas.""",

    """Reading ten is a transcript regarding the first people of Sulu, showing early cultural understanding, though largely mythical.""",

    """Excerpts from “A New Species of Homo in the Late Pleistocene of the Philippines” document the discovery of the Right Metatarsal 3 (RMT3) in Callao Cave, dated to 67,000 years ago, and comparisons to Southeast Asian human remains.""",

    """Excerpts from “Upper Pleistocene Homo sapiens From the Tabon Cave” describe the reassessment of Tabon Man remains, including new dating and eleven additional human fossils, helping fill the Upper Pleistocene gap in the Philippines.""",

    """Excerpts from “The Austronesians: Historical and Comparative Perspective” provide background on Austronesian expansion, linguistic evidence, and migration patterns from Taiwan through the Philippines, Borneo, Sulawesi, Java, Sumatra, Malaya, and Oceania."""
]

# ---- Step 3: Generate embeddings ----
domain_embedding = model.encode([domain_description])
chunk_embeddings = model.encode(chunks)

# ---- Step 4: Compute cosine similarity ----
similarities = cosine_similarity(domain_embedding, chunk_embeddings)

# ---- Step 5: Display results ----
print("Domain Similarity Scores:\n")

for chunk, score in zip(chunks, similarities[0]):
    print(f"Score: {score:.4f}")
    print(f"Chunk: {chunk}")
    print("-" * 60)
```
